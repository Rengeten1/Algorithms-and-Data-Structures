Review the list implementation in Algorithm 2. Extend it by implementing setitem (self, k,
value), insert(self, k, value) and remove(self, k) methods.

```python
import ctypes # provides make_array function

class MyArray:
    def __init__(self):
        self._n = 0 # number of elements
        self._capacity = 1 # array capacity 
        self._A = self._make_array(self._capacity) # create array of capacity 1

    # returns the number of elements in the array
    def __len__(self):
        return self._n

    # gets the element at index k
    def __getitem__(self, k):
        if not 0 <= k < self._n:
            raise IndexError('invalid index')
        return self._A[k]

    # adds an element to the end of the array
    def _append(self, obj):
        if self._n == self._capacity:
            self._resize(2 * self._capacity)
        self._A[self._n] = obj
        self._n += 1

    # resizes the array to capacity c
    def _resize(self, c):
        B = self._make_array(c)
        for k in range(self._n):
            B[k] = self._A[k]
        self._A = B
        self._capacity = c

    # creates a new array of capacity c
    def _make_array(self, c):
        return (c * ctypes.py_object)()
    
    # sets the element at index k to value
    def _setitem(self, k, value):
        if not 0 <= k < self._n:
            raise IndexError('Invalid Index')
        self._A[k] = value

    # inserts an element
    def _insert(self, k, value):
        if not 0 <= k < self._n:
            raise IndexError('Invalid Index')
        if self._n == self._capacity:
            self._resize(2 * self._capacity)

    # removes an element at index k
    def _remove(self, k):
        if not 0 <= k < self._n:
            raise IndexError('Invalid Index')
        for j in range(k, self._n - 1):
            self._A[j] = self._A[j + 1]
        self._A[self._n - 1] = None
        self._n -= 1
        if self._n < self._capacity // 4:
            self._resize(self._capacity // 2)
```

The syntax data.remove(value) for Python list data removes only the first occurrence of element value from the list. Give an implementation of a function, with signature remove all(data, value), that removes all occurrences of value from the given list, such that the worst-case running time of the function is O(n) on a list with n elements. Not that it is not efficient enough in general to rely on repeated calls to remove.

```python
# remove method removes the first element of the array with the value 

# push everything to the end of the array and then del everything after the write
def all(data, value):
    if data == []:
        raise IndexError('Cant remove from empty array')
    write = 0
    for read in range(len(data)):
        if data[read] != value:
            data[write] = data[read]
            write += 1

    del data[write:]
```

```python
a = []
for i in range(10):
    a.append(i)
    a.append(1)

print('----pre-----')
for i in range(len(a)):
    print(a[i])

all(a, 1)

print('----post-----')
for i in range(len(a)):
    print(a[i])
```

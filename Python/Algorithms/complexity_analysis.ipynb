Graph the functions $8n$, $4 n log(n)$, $2n^2$, $n3$, and $2n$ using a logarithmic scale for the $x$-axis and $y$-axes; that is, if the function value $f(n)$ is $y$, plot this as a point with $$x$-coordinate at $log(n)$ and $y$-coordinate at $log(y)$.

```python
import matplotlib.pyplot as plt
import numpy as np

def plot_functions():
    n = np.arange(1, 1000, 1)
    functions = {
        '8n': 8 * n,
        '4n log n': 4 * n * np.log(n),
        '2n^2': 2 * n**2,
        'n^3': n**3,
        '2^n': 2**n
    }

    plt.figure(figsize=(10, 6))

    for label, values in functions.items():
        plt.plot(np.log(n), np.log(values), label=label)

    plt.xlabel('log(n)')
    plt.ylabel('log(f(n))')
    plt.title('Logarithmic Plot of Functions')
    plt.legend()
    plt.grid(True)
    plt.show()

plot_functions()
```

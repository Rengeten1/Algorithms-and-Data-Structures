#### Bubble Sort
Bubble sort is a simple sorting algorithm that repeatedly goes through the list, compares and swaps the adjacent elements if they are in the wrong order. It continues until no swaps are needed, which means the list is sorted.


```python
def bubble_sort(a):
    n = len(a)
    for i in range(n):
        for j in range(n-i-1):
            if a[j] > a[j+1]:
                tmp = a[j]
                a[j] = a[j+1]
                a[j+1] = tmp
    return a


a = [10, 2, 4 , 6, 1, 4, 2, 5]
bubble_sort(a)

%timeit bubble_sort([1, 2, 2, 4, 4, 5, 6, 10])
```

First We implement the simple bubble sort algorithm:

```python
def bubble_sort_improved(a):
    n = len(a)

    for i in range(n):
        swapped = False
        for j in range(n-i-1):
            if a[j] > a[j+1]:
                tmp = a[j]
                a[j] = a[j+1]
                a[j+1] = tmp 
                swapped = True
        if not swapped:
            break

    return a

%timeit bubble_sort_improved([1, 2, 2, 4, 4, 5, 6, 10])
```

#### Merge Sort
Merge sort is a divide and conquer algorithm that divides the input array into halves, sorts them recursively, and then merges the sorted halves. 

```python
def merge(S1, S2, S):
    """
    Merge two sorted arrays S1 and S2 into array S.
    
    Args:
        S1: First sorted subarray
        S2: Second sorted subarray
        S: Original array to store the merged result
    """
    i = j = 0
    n, m = len(S1), len(S2)
     
    while i < n and j < m:
        if S1[i] <= S2[j]:
            S[i + j] = S1[i] 
            i += 1
        else:
            S[i+j] = S2[j]
            j += 1
    # adding unassigned elements
    while i < n:
        S[i + j] = (S1[i])
        i += 1

    while i < n:
        S[i +j] =  (S2[j])
        j += 1
```

```python
def merge_sort(S):
    n = len(S)
    if n < 2:
        return
    mid = n//2
    S1 = S[:mid]
    S2 = S[mid:]
    merge_sort(S1)
    merge_sort(S2)
    merge(S1, S2, S)
    return S
```

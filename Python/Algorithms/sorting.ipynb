#### Stability of Sorting Algorithms
A sorting algorithm is said to be stable if two objects with equal keys appear in the same order in sorted output as they appear in the input data set

#### Bubble Sort
Bubble sort is a simple sorting algorithm that repeatedly goes through the list, compares and swaps the adjacent elements if they are in the wrong order. It continues until no swaps are needed, which means the list is sorted.


```python
def bubble_sort(a):
    n = len(a)
    for i in range(n):
        for j in range(n-i-1):
            if a[j] > a[j+1]:
                tmp = a[j]
                a[j] = a[j+1]
                a[j+1] = tmp
    return a


a = [10, 2, 4 , 6, 1, 4, 2, 5]
bubble_sort(a)

%timeit bubble_sort([1, 2, 2, 4, 4, 5, 6, 10])
```

The bubble sort implementation we discussed in class has a best-case running time complexity of O(n2). Implement a better bubble sort that has a best-case running time complexity of O(n). Hint: stop as early as possible, i.e. if the sequence was not altered at all in the previous iteration.

First We implement the simple bubble sort algorithm:

```python
def bubble_sort_improved(a):
    n = len(a)

    for i in range(n):
        swapped = False
        for j in range(n-i-1):
            if a[j] > a[j+1]:
                tmp = a[j]
                a[j] = a[j+1]
                a[j+1] = tmp 
                swapped = True
        if not swapped:
            break

    return a

%timeit bubble_sort_improved([1, 2, 2, 4, 4, 5, 6, 10])
```

#### Merge Sort
Merge sort is a divide and conquer algorithm that divides the input array into halves, sorts them recursively, and then merges the sorted halves. 

```python
def merge(S1, S2, S):
    i = j = 0
    while i < len(S1) and j < len(S2):
        if S1[i] <= S2[j]:
            S.append(S1[i])
            i += 1
        else:
            S.append(S2[j])
            j += 1

    while i < len(S1):
        S.append(S1[i])
        i += 1

    while j < len(S2):
        S.append(S2[j])
        j += 1
```

```python
def merge_sort(S):
    n = len(S)
    if n < 2:
        return
    mid = n//2
    S1 = S[:mid]
    S2 = S[mid:]
    merge_sort(S1)
    merge_sort(S2)
    merge(S1, S2, S)
    return S
```

Implement merge(S1, S2, S) for merge sort

```python
def merge(S1, S2, S):
    i = j = k = 0
    while i < len(S1) and j < len(S2):
        if S1[i] <= S2[j]:
            S[k] = S1[i]
            i += 1
        else:
            S[k] = S2[j]
            j += 1
        k += 1

    while i < len(S1):
        S[k] = S1[i]
        i += 1
        k += 1

    while j < len(S2):
        S[k] = S2[j]
        j += 1
        k += 1
```

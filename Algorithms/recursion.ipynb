# Recursion
Recursion is a programming technique where a function calls itself to solve a smaller instance of the same problem. It typically involves a base case to terminate the recursion and a recursive case that breaks the problem down into smaller subproblems.

**Best case time complexity:** *O(n)*

**Average case time complexity:** *O(n)*

**worst case time complexity:** *O(n)*

Describe a recursive algorithm for finding the maximum element in a sequence, S, of n elements. What is your running time and space usage?

```psuedo
def recursive_max(S):
    if len(S) == 1:
        return S[0]
    else:
        max_of_rest = recursive_max(S[1:])
        return max_of_rest if max_of_rest > S[0] else S[0]
```
The running time of this algorithm is $O(n)$, and the space usage is $O(n)$ due to the recursive call stack.

````python
def recursive_max_num(S):
    if not S:
        return 0
    elif len(S) == 1:
        return S[0]
    else:
        current = recursive_max_num(S[1:])
        return current if current > S[0] else S[0]
    
l = list(range(1000))

recursive_max_num(l)
````

Write a short recursive Python function that finds the minimum and maximum values in a sequence without using any loops

````python
# Write a short recursive Python function that finds the minimum and maximum values in a sequence without using any loops
def recursive_max_min(s):
    if not s:
        return None, None
    elif len(s) == 1:
        return s[0], s[0]
    else:
        min_val, max_val = recursive_max_min(s[1:])
        return (min_val if min_val < s[0] else s[0],
                max_val if max_val > s[0] else s[0])
````

Write a short recursive Python function that determines if a string s is a palindrome, that is, it is equal to its reverse. For example, racecar and gohangasalamiimalasagnahog are palindromes.

````python
def palindrome_check(string):
    if len(string) < 1:
        return True
    elif string[0] != string[-1]:
        return False
    else:
        return palindrome_check(string[1:-1])

palindrome_check('racecar')
palindrome_check('gohangasalamiimalasagnahog')
````

A trivial recursive definition of the power function follows from the fact that $x^n = x × x^{n−1}$ for $n > 0$

$power(x, n) =
 1  ,                   n = 0 |
  x × power(x, n − 1) , n > 0 |

Implement this function and determine its running time complexity

````python
def power(x, n):
    if n == 0:
        return 1
    else:
        return 2 * power(x, n-1)
    
# the running time complexity of this will be 
# O(n) + O(n-1) + ... + O(1) = O(n²)
````

There is a much faster way to compute the power function using an alternative definition that employs a squaring technique:

\
$power(x, n)$ = 
\begin{cases}
1 & \text{if } n = 0 \\
\left( \text{power}\left(x, \left\lfloor \frac{n}{2} \right\rfloor \right) \right)^2 & \text{if } n > 0 \text{ and } n \text{ is even} \\
x \times \left( \text{power}\left(x, \left\lfloor \frac{n}{2} \right\rfloor \right) \right)^2 & \text{if } n > 0 \text{ and } n \text{ is odd}
\end{cases}


Let $\left\lfloor \frac{n}{2} \right\rfloor$ denote the floor of the division (expressed as `n // 2` in Python).

**Implement this function and determine its running time complexity.**

````python
def power(x, n):
    if n == 0:
        return 1
    elif n%2 :
        return x * (power(x, n//2)) **2
    else:
        return  (power(x, n//2)) ** 2  

power(5, 1)

# running time complexity of O(log n)
````

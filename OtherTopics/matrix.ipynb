Fast Matrix Multiplication
========================
The runtime for multiplying rectangular matrices (m x n) and (n x p) is $O(mnp)$. This means that the time it takes to multiply two matrices grows with the product of their dimensions.

The normal matrix multiplication algorithm has a time complexity of $O(n^3)$. However, there are more efficient algorithms that can perform matrix multiplication in less time.


One of the most famous algorithms is Strassen's algorithm, which has a time complexity of approximately O(n^2.81).

# Strassen's Algorithm
Let $A, B$ be two matrices of size $n x n$, we want to compute the product $C = A * B$. if $A, B$ are not of type $2^k * 2^k$, we can fill the missing rows and columns with zeros to make them of size $2^k * 2^k$.

## example
Let $A = \begin{bmatrix} a1,1 & a1,2 \\ a2,1 & a2,2 \end{bmatrix}$,  $B = \begin{bmatrix} b1,1 & b1,2 \\ b2,1 & b2,2 \end{bmatrix}$ and $C = \begin{bmatrix} c1,1 & c1,2 \\ c2,1 & c2,2 \end{bmatrix}$ 

we can compute the product $C = A * B$ as follows:

$$C1.1 = a1,1 * b1,1 + a1,2 * b2,1$$

$$C1.2 = a1,1 * b1,2 + a1,2 * b2,2$$

$$C2.1 = a2,1 * b1,1 + a2,2 * b2,1$$

$$C2.2 = a2,1 * b1,2 + a2,2 * b2,2$$

Now by using Strassen's algorithm, we can compute the product $C$ in a more efficient way, by instead of computing the 8 products, we can compute only 7 products as follows:

$$M1 = (a1,1 + a2,2) * (b1,1 + b2,2)$$

$$M2 = (a2,1 + a2,2) * b1,1$$

$$M3 = a1,1 * (b1,2 - b2,2)$$

$$M4 = a2,2 * (b2,1 - b1,1)$$

$$M5 = (a1,1 + a1,2) * b2,2$$

$$M6 = (a2,1 - a1,1) * (b1,1 + b1,2)$$

$$M7 = (a1,2 - a2,2) * (b2,1 + b2,2)$$

Now we can compute the product $C$ as follows:

$$C1.1 = M1 + M4 - M5 + M7$$

$$C1.2 = M3 + M5$$

$$C2.1 = M2 + M4$$

$$C2.2 = M1 - M2 + M3 + M6$$

We iterate this process n times until the submatrices degenerate into numbers. The resulting product will be padded with zeros like A and B, and should be stripped of the corresponding rows and columns to get the final product.

- The running time of Strassen's algorithm is $O(n^{\log_2 7}) \approx O(n^{2.81})$, which is faster than the normal matrix multiplication algorithm.

- Because of a large constant, it often only used for matrices larger than for n > 100.

- Drawback : It requires more memory to store the intermediate matrices.
   - The __numerical stability__ is reduced compared to the naive algorithm.

- There are even more efficient algorithms, but they come with a constant that is so large that these algorithms are only worthwhile for matrices that are too large to handle on 
present-day computers.

The problem is not actually to perform the multiplications, but merely to decide in which order to perform the multiplications. We have many options to multiply a chain of matrices because matrix multiplication
is associative. In other words, no matter how we parenthesize the product, the result will be the same. For example, if we had four matrices A, B, C and D, we would have:

$(ABC)D = (AB)(CD) = A(BCD) = ....$

However, the order in which we parenthesize the product affects the number of simple multiplications needed to compute the product, or the efficiency. For example, suppose $A$ is a $10 × 30$ matrix, $B$ is a $30 × 5$ matrix and $C$ is a $5 × 60$ matrix. 

Then $(AB)C = (10 × 30 × 5) + (10 × 5 × 60) = 1500 + 3000 = 4500$ multiplications.

How many multiplications does it take to compute $A(BC)$? Which of the two parenthesizations requires less number of multiplications?
